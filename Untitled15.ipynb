```python
# Instalação das bibliotecas
!pip install -q transformers datasets accelerate scikit-learn pandas torch

import pandas as pd
import json
import torch
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    Trainer,
    TrainingArguments,
    EarlyStoppingCallback
)
from sklearn.metrics import accuracy_score
from datasets import Dataset

# 1. Carregar os dados
def load_jsonl(file_path):
    data = []
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            data.append(json.loads(line))
    return pd.DataFrame(data)

train_df = load_jsonl('train.jsonl')
val_df = load_jsonl('validation.jsonl')
test_df = pd.read_csv('test.csv')

train_df['label'] = train_df['label'].astype(int)
val_df['label'] = val_df['label'].astype(int)

# 2. Tokenização com BERTimbau Large
model_name = "neuralmind/bert-large-portuguese-cased"
tokenizer = AutoTokenizer.from_pretrained(model_name)

def tokenize_function(examples):
    return tokenizer(examples['text'], padding="max_length", truncation=True, max_length=128)

train_ds = Dataset.from_pandas(train_df[['text', 'label']])
val_ds = Dataset.from_pandas(val_df[['text', 'label']])
test_ds = Dataset.from_pandas(test_df[['text']])

train_tokenized = train_ds.map(tokenize_function, batched=True)
val_tokenized = val_ds.map(tokenize_function, batched=True)

# 3. Configurar o Modelo
model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=2)

# 4. Configurar o Treinamento (Ajustado para performance máxima)
training_args = TrainingArguments(
    output_dir='./results',
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=1e-5,                     # Taxa reduzida para aprendizado mais estável
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=6,                     # Mais margem para o modelo aprender com segurança
    weight_decay=0.01,
    warmup_ratio=0.1,                       # Aquecimento para estabilizar os pesos iniciais
    load_best_model_at_end=True,
    metric_for_best_model="accuracy",       # Otimiza focado na acurácia real
    greater_is_better=True,                 # Maior acurácia é melhor
    push_to_hub=False,
    report_to="none"
)

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = logits.argmax(axis=1)
    acc = accuracy_score(labels, preds)
    return {"accuracy": acc}

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_tokenized,
    eval_dataset=val_tokenized,
    compute_metrics=compute_metrics,
    callbacks=[EarlyStoppingCallback(early_stopping_patience=2)] # Tolerância aumentada para evitar paradas precipitadas
)

# 5. Treinar a IA
trainer.train()

# 6. Gerar previsões e salvar o arquivo de submissão
test_tokenized = test_ds.map(tokenize_function, batched=True)
predictions = trainer.predict(test_tokenized)
preds = predictions.predictions.argmax(axis=1)

submission = pd.DataFrame({
    'id': test_df['id'],
    'label': preds
})

submission.to_csv('submission.csv', index=False)
print("Pronto! Arquivo 'submission.csv' gerado com sucesso!")
```
